# **Estudo de funções**

Funções são **blocos de código reutilizáveis** que executam uma tarefa específica.

- Sua sintaxe é definida usando a palavra-chave `def`, seguida pelo `nome da função` e parênteses que podem conter `parâmetros`. O bloco de código da função é **identado** e pode conter um retorno opcional com o `return`.

```
def nome_da_função(parâmetros):
    # Bloco de código da função
    return valor_opcional
```



As funções podem váriar de acordo com o contexto.

- **Parâmetros** são variáveis definidas na declaração da função, elas podem ser obrigatórias ou opcionais, sendo usadas para tornar as funções mais flexíveis, permitindo que elas processem diferentes valores sem precisar modificar o código.

````python
def ola_mundo(nome):
    return f"Olá, {nome}!"
````

Para chamar uma função, basta usar o **nome da função seguido de parênteses**.

Argumentos são os **valores passados para a função quando ela é chamada**, caso essa função tenha um parâmetro obrigatório.

````python
ola_mundo("Laís")
````

## **Funções simples**

Estas são as funções criadas e personalizadas pelo dev para realizar tarefas específicas.

A função `somar()` recebe dois parâmetros (a e b) e retorna a soma desses valores.

O bloco de execução e as variáveis criadas dentro de uma função são locais, ou seja, **só existem dentro dela**. Isso significa que elas não podem ser acessadas fora do seu bloco de código.

**O que é**: Uma máquina que recebe ingredientes (parâmetros), processa e devolve um bolo (retorno).

````python
def somar(a, b):
  soma = a + b
  return soma
````

````python
print(somar('marcela', ' e jorge'))
somar(13.5, 7.8)
````

**Quando usar**: Em 90% dos casos. Use sempre que tiver um trecho de código que se repete ou para organizar o raciocínio. Se precisa calcular e devolver um dado, use return.

## **Funções com parâmetros padrão**

Estas são as funções que permitem definir valores padrão para os parâmetros caso nenhum argumento seja passado.

**O que é**: Uma função "educada". Se você não disser o que quer, ela já tem uma sugestão pronta.

````python
def cumprimentar(nome = 'Visitante'):
  print(f'Olá, {nome}!')
````

Se nenhum nome for passado, **o padrão será "Visititante"**. Se um nome for fornecido, ele será usado na saudação.

````python
cumprimentar() # Saída: Olá, Visitante!

cumprimentar('Jade') # Saída: Olá, Jade!
````

**Quando usar**: Quando um parâmetro é opcional.

*Exemplo*: Uma função de "Buscar Clientes" onde o filtro de cidade é opcional. Se não passar nada, busca de todas as cidades.

## **Funções anônimas (lambdas)**

Lambda é uma função pequena e sem nome que pode ter múltiplos argumentos, mas apenas uma expressão.

Lambdas são úteis para operações simples ou quando precisamos de funções curtas dentro de outras funções.

**O que é**: É como uma fórmula matemática rápida de excel. Você não quer criar toda a estrutura de uma função para algo que cabe em uma linha.

````python
# Sintaxe: lambda argumentos: expressão

soma = lambda a, b: a + b
````

````python
print(soma(3, 5)) # Saída: 8
````

1. **Ordenação Customizada (Onde elas são mais usadas)**

Imagine que você tem uma lista de dicionários e quer ordenar pelo preço. Sem `lambda`, você teria que criar uma função def só para isso.

````python
produtos = [
    {"nome": "Teclado", "preco": 150},
    {"nome": "Mouse", "preco": 80},
    {"nome": "Monitor", "preco": 900}
]

# Ordenar pelo preço (do menor para o maior)
produtos_ordenados = sorted(produtos, key=lambda p: p["preco"])

def ordenar(produtos):
  produtos_ordenados = sorted(produtos, key=lambda p: p['preco'])
  return produtos_ordenados
````

````python
print(produtos_ordenados)
ordenar(produtos)
````

2. **Filtragem de Dados com filter()**

A função `filter` extrai itens de uma lista que atendem a uma condição.

````python
idades = [12, 18, 25, 105, 40, 14, 30, 8, 29]

# Queremos apenas as idades maiores ou iguais a 18
maiores_de_idade_padrao = list(filter(lambda x: x >= 18, idades))
# Queremos apenas as idades maiores ou iguais a 18 em ordem crescente
maiores_de_idade_ordenado= sorted(list(filter(lambda x: x >= 18, idades)))
print(f'Lista ordenada: {maiores_de_idade_ordenado}')
print(f'Lista padrão: {maiores_de_idade_padrao}')
````

3. **Transformação de Dados com map()**

A função `map` aplica uma regra a todos os itens de uma lista.

````python
precos_em_reais = [90, 256, 450, 55, 99, 89, 366]

# Aplicar um aumento de 150% em todos os preços
novos_precos = list(map(lambda p: p + p * 0.5, precos_em_reais))

print(f"Preços normais: {precos_em_reais}")
# Fórmula de porcentagem de aumento de um valor referente a ele mesmo antes: (Valor Final - Valor Inicial) / Valor Inicial * 100
print(f'Preços com 50% de aumento: {novos_precos}') # Saída: [110.0, 275.0, 440.0]
````

4. **Usando Condicionais (Ternários) dentro da Lambda**

Embora não possa usar um bloco if completo, você pode usar a estrutura de "if de uma linha".

````python
# Verifica se um número é par ou ímpar
par_ou_impar = lambda x: "Par" if x % 2 == 0 else "Ímpar"

print(f'O número é {par_ou_impar(7)}')  # Saída: Ímpar
print(f'O número é {par_ou_impar(10)}') # Saída: Par
````

#### **Quando NÃO usar Lambdas?**

**Se a lógica for complexa:** Se você precisar de mais de uma linha, use def.

**Se você for reutilizar muito:** Funções nomeadas com def são melhores para documentação e reuso em diferentes partes do projeto.

> **Dica de Ouro:** Pense na lambda como uma "função descartável". Você a cria ali na hora, ela executa o que precisa e pronto.

## **Funções recursivas**

São aquelas funções que **chamam a si mesmas** para resolver problemas repetitivos até atingir uma condição de parada.

Elas são usadas para resolver problemas que podem ser divididos em subproblemas menores, como cálculos matemáticos e algoritmos de estrutura de dados.

````python
def fatorial(n):
  if n == 0: # Condição de parada
    return 1
  return n * fatorial(n - 1) # Chamada recursiva
````

````python
print(fatorial(5)) # Saída: 120
````

### **A Anatomia de uma Função Recursiva**

Para que uma função recursiva não rode para sempre (o que causaria um erro de "estouro de pilha" ou Stack Overflow), ela precisa de duas partes obrigatórias:

> **Caso Base (Condição de Parada)**: É o momento em que a função para de chamar a si mesma. No seu exemplo, é o if n == 0: return 1. Sem isso, o computador ficaria sem memória tentando empilhar infinitas chamadas.

> **Passo Recursivo:** É onde a função chama a si mesma, mas com um problema ligeiramente menor. No exemplo: n * fatorial(n - 1). Note que o valor de n diminui a cada chamada, aproximando-o do zero (o caso base).

#### **Como o Python processa o Fatorial(3)?**

Para visualizar o que acontece "por baixo dos panos" quando você chama fatorial(3), imagine uma pilha de tarefas:

- Chamada 1: fatorial(3) → O Python vê: 3 * fatorial(2). (Ele "pausa" e vai resolver o fatorial de 2).
- Chamada 2: fatorial(2) → O Python vê: 2 * fatorial(1). (Pausa novamente).
- Chamada 3: fatorial(1) → O Python vê: 1 * fatorial(0). (Pausa novamente).
- Chamada 4 (Caso Base): fatorial(0) → Retorna 1.

> Agora a pilha começa a "descer" resolvendo as contas:

- A Chamada 3 recebe o 1 e faz: $1 \times 1 = 1$.
- A Chamada 2 recebe o 1 e faz: $2 \times 1 = 2$.
- A Chamada 1 recebe o 2 e faz: $3 \times 2 = 6$.
- Resultado Final: 6.

#### **Outro Exemplo: Contagem Regressiva**

Nem toda recursão precisa de cálculos matemáticos complexos. Veja este exemplo de uma contagem regressiva:

````python
def contagem_regressiva(n):
    if n == 0:
        print("BUM! 🚀") # Caso Base
    else:
        print(n)
        contagem_regressiva(n - 1) # Chamada Recursiva

contagem_regressiva(3)
# Saída:
# 3
# 2
# 1
# BUM! 🚀
````

#### **Quando usar e quando evitar?**

> **Use quando:** O problema for naturalmente divisível em partes menores idênticas (como percorrer pastas e subpastas de um computador ou estruturas de dados como Árvores).

> **Evite quando:** O problema for muito simples e puder ser resolvido com um loop for ou while comum. Em Python, a recursividade tem um limite (geralmente 1000 chamadas) e consome mais memória que um loop simples.

## **Funções dentro de funções (Closures)**

Closure é uma **função interna** que lembra do ambiente onde foi criada, mesmo após execução da função externa ter terminado. Isso permite que ela mantenha o estado de variáveis sem a necessidade de usá-las globalmente.

````python
def multiplicador(n): # Função externa
  def multiplica(x): # Closure
    return x * n # usa a variável 'n' da função externa
  return multiplica # Retorna a função interna
````

As `closures` são úteis para encapsular lógicas e criar funções especializadas sem usar variáveis globais.

````python
triplo = multiplicador(3)
valor = triplo(5)
print(valor) # Saída: 15
````

````python
dez_vezes = multiplicador(10)
valor = dez_vezes(3)

print(valor)
````

#### **Como uma Closure funciona na prática?**

Para algo ser considerado uma Closure, ele precisa seguir três regras:

- Ter uma função dentro de outra função (aninhada).

- A função interna deve usar uma variável que pertence à função externa.

- A função externa deve retornar a função interna.

````python
def criar_saudacao(saudacao):
  def saudar(nome):
    return f"{saudacao}, {nome}!"
  return saudar
````

````python
bom_dia = criar_saudacao('Bom dia')
boa_noite = criar_saudacao('Boa noite')

print(bom_dia("Ana"))
print(boa_noite("William"))
````

**Exemplo: O Contador de Pontos**

Imagine que você está criando um jogo e quer manter o placar de cada jogador sem usar variáveis globais (que podem ser bagunçadas).

````python
def criar_placar():
    pontos = 0  # Esta variável está no "ambiente" da função externa

    def adicionar_ponto():
        nonlocal pontos # Avisa ao Python que queremos usar a variável de "fora"
        pontos += 1
        return pontos

    return adicionar_ponto # Retorna a função SEM os parênteses

# Criando instâncias independentes
placar_jogador_1 = criar_placar()
placar_jogador_2 = criar_placar()

print(placar_jogador_1()) # Saída: 1
print(placar_jogador_1()) # Saída: 2
print(placar_jogador_2()) # Saída: 1 (O estado é independente!)
````

#### **Por que usar Closures?**

As principais vantagens são:

> **Manter o Estado:** Como no exemplo do placar, a função interna mantém o valor de pontos vivo, mesmo depois que criar_placar() já terminou de executar.

> **Encapsulamento (Privacidade):** A variável pontos está protegida. Ninguém fora da função pode alterá-la diretamente; ela só pode ser modificada através da função adicionar_ponto.

> **Substituir Classes Simples:** Às vezes, criar uma classe inteira para guardar apenas uma variável é exagero. Uma closure resolve isso de forma muito mais leve e rápida.

**Parâmetros padrão (default parameters)**

Tornam as funções mais flexíveis. No backend, isso é a base da **Paginação de Dados**.

Imagine que você tem uma rota na sua API que lista produtos. Se o usuário não disser quantos quer ver, você precisa ter um padrão para não travar o banco de dados devolvendo 1 milhão de itens.

````python
# Exemplo Prático: Paginação de API
def listar_produtos(pagina=1, limite_por_pagina=20):
    # Simulação de cálculo de offset para banco de dados
    inicio = (pagina - 1) * limite_por_pagina
    fim = inicio + limite_por_pagina

    return f"Buscando no DB: registros de {inicio} até {fim}"

# Chamadas
print(listar_produtos())            # Usa o padrão: página 1, 20 itens [cite: 38]
print(listar_produtos(pagina=3))    # Usa página 3, limite padrão de 20
````

`map` e `filter` com listas de números. No backend, 90% do tempo você lidará com listas de dicionários (que viram JSONs).

- Digamos que você **recebeu uma lista de usuários do banco de dados** e precisa **filtrar** apenas os **ativos** e **extrair apenas os e-mails** para enviar uma newsletter.

Isso é muito mais rápido e limpo do que criar loops **for** vazios e ficar dando `.append()`.

````python
usuarios = [
    {'id': 1, 'nome': 'Will', 'ativo': True, 'email': 'will@email.com'},
    {'id': 2, 'nome': 'Ana', 'ativo': False, 'email': 'ana@email.com'},
    {'id': 3, 'nome': 'Pedro', 'ativo': True, 'email': 'pedro@email.com'}
]

# 1. Filter: Pegar apenas usuários ativos [cite: 149]
# Usamos lambda para verificar a chave 'ativo'
ativos = list(filter(lambda u: u['ativo'] == True, usuarios))

# 2. Map: Extrair apenas o email dos ativos [cite: 150]
emails_para_envio = list(map(lambda u: u['email'], ativos))
print(emails_para_envio)
# Saídaill@email.com', 'pedro@email.com']
````

**Closures**, funções dentro de funções que lembram o ambiente onde foram criadas. No Python Backend (Django, Flask, FastAPI), as closures são a base dos **Decorators** `(@)`.

Um Decorator nada mais é que uma closure usada para **adicionar funcionalidade a uma função "de fora", sem mexer no código dela**.

**Cenário**: Verificar se um usuário está logado antes de executar uma função sensível.

````python
def requer_autenticacao(funcao_original):
    # Esta é a closure (função interna)
    def wrapper(usuario):
        if not usuario.get('logado'):
            return "Acesso Negado: Faça login."
        return funcao_original(usuario) # Executa a função se passar no teste

    return wrapper [224]

# Simulando o uso com a sintaxe de Decorator (@)
@requer_autenticacao
def ver_painel_financeiro(usuario):
    return f"Bem-vindo ao painel, {usuario['nome']}! Saldo: R$ 10.000"

# Teste
user_will = {'nome': 'Will', 'logado': True}
user_anonimo = {'nome': 'Visitante', 'logado': False}

print(ver_painel_financeiro(user_will))     # Acesso liberado
print(ver_painel_financeiro(user_anonimo))  # Acesso Negado
````

### ***args** e  ****kwargs**
 parâmetros tornam funções flexíveis. O *args e **kwargs são a forma suprema disso. Eles permitem passar um número indefinido de argumentos.

`*args`: Recebe múltiplos argumentos como uma Tupla.

`**kwargs`: Recebe múltiplos argumentos nomeados como um Dicionário (Keyword Arguments).

````python
def salvar_log(*mensagens, **detalhes):
    for msg in mensagens:
        print(f"LOG: {msg}")

    if 'nivel' in detalhes:
        print(f"Nível de severidade: {detalhes['nivel']}")
        print(f'Data: {detalhes['data']}')
````

````python
salvar_log("Erro de conexão", "Tentativa 2 falhou", nivel="CRÍTICO", data="17/11")
````

### **enumerate()**
Sempre que você precisar percorrer uma lista e também precisar do índice (a posição) do item, não use contadores manuais. Use `enumerate`.

O objeto enumerate retorna pares contendo uma contagem (a partir de start, que
tem como padrão zero) e um valor retornado pelo argumento iterável.

enumerate é útil para obter uma lista indexada:

(0, seq[0]), (1, seq[1]), (2, seq[2]), ...

````python
lista_tarefas = ["Estudar Python", "Ler PDF", "Praticar"]

for indice, tarefa in enumerate(lista_tarefas):
    print(f"{indice + 1} - {tarefa}")
````

### **any() e all()**
São funções built-in ótimas para validação de formulários ou regras de negócio.

`all()`: Retorna True se todos os itens forem verdadeiros.

`any()`: Retorna True se pelo menos um item for verdadeiro.

````python
requisitos_senha = [True, True, False] # Ex: Tem maiúscula? Tem número? Tem símbolo?

if not all(requisitos_senha):
    print("Sua senha não atende a todos os requisitos.")
````

````python
requisitos_senha = [False, False, False] # Ex: Tem maiúscula? Tem número? Tem símbolo?

if not any(requisitos_senha):
  print('Sua senha não atende a nenhum dos requisitos.')
else:
  print('Sua senha atende a pelo menos um dos requisitos.')
````

O ***args** (o nome pode ser qualquer um, o que importa é o asterisco *) **empacota tudo o que você mandar numa tupla**.

**Cenário**: Uma calculadora que soma quantos números você quiser.

````python
def somar_tudo(*numeros):
    # O Python transforma os números numa tupla: (10, 20, 30, ...)
    print(f"Números recebidos: {numeros}")
    return sum(numeros) # Usa a função sum() que você listou no PDF
````

````python
# Testando com quantidades diferentes
print(f'Soma da tupla: {somar_tudo(10, 20)}')          # Saída: 30
print(f'Soma da tupla: {somar_tudo(1, 2, 3, 4, 5)}')   # Saída: 15
````

Entendendo ****kwargs**: A "Ficha de Cadastro":

Enquanto o ***args** lida com listas de **valores soltos**, o ****kwarg** lida com **valores nomeados** (chave e valor). Pense nele como preencher campos opcionais num formulário.

**Cenário**: Criar um perfil de usuário onde apenas o nome é obrigatório, mas o resto (idade, cidade, profissão) é opcional.

````python
def criar_perfil(nome, **dados_opcionais):
  if len(nome) < 4:
    print('Impossível criar o perfil. Não contém nome.')
  else:
    return f'Nome: {nome}' '\n' f'Dados extras: {dados_opcionais}'
````

````python
perfil_01 = criar_perfil('William Coelho', idade= 29, cidade='Recife', estudante=True)

print(perfil_01)
````

### **Diferença entre `print()` e return**

**A Regra de Ouro**

`print()` é para o **Humano ver**: Ele serve apenas para mostrar algo na tela (console). O computador **"esquece" o valor logo depois de mostrar**.


`return` é para o **Computador usar**: Ele entrega o valor para quem chamou a função, para que o programa continue usando esse dado depois.

### **Analogia do caixa eletrônico**:

Imagine que você vai ao banco sacar dinheiro.

- Função com print() (O Recibo): A máquina imprime um papel dizendo: "Seu saldo é R$ 100,00".

Você consegue comprar pão com esse papel? Não. É apenas uma informação visual.

- Função com return (O Dinheiro): A máquina cospe uma nota de R$ 100,00.

Você consegue comprar pão com isso? Sim. Você pegou o valor e pode usá-lo em outra "operação" (a padaria).

- **Erro Clássico** (Visualizando no Código): Olhe o exemplo da função somar usando return. Vamos ver o que acontece se trocarmos por print.

````python
def somar_errado(a, b):
    resultado = a + b
    print(resultado) # Mostra na tela, mas NÃO devolve o valor

# Tentar usar o resultado numa conta
valor = somar_errado(10, 10)
# O Python mostra "20" na tela agora.
# Mas a variável 'valor' está vazia (None).

# O código abaixo VAI QUEBRAR:
calculo_final = valor + 5
# Erro! Porque você está tentando somar 5 com "Vazio".
````

````python
def somar_certo(a, b):
    resultado = a + b
    return resultado # Entrega o valor 20 para quem pediu

# Tentar usar o resultado numa conta
valor = somar_certo(10, 10)
# Nada aparece na tela agora, o 20 está guardado na variável 'valor'.

# O código abaixo FUNCIONA:
calculo_final = valor + 5
print(calculo_final)
# Saída: 25
````
