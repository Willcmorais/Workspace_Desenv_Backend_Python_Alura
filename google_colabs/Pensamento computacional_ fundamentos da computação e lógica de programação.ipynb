# O pensar computacional

É a habilidade de conseguir abstrair um problema, criando um **passo a passo bem explicado** para que o computador consiga resolver o problema demandado da melhor forma possível: `rápida`, `confiável`, `de fácil utilização pelo usuário`, `que seja de fácil manutenção` e tenha `escalabilidade`.

Computação é um conjunto de sistemas que interagem entre sí para entregar uma certa demanda pedida. Para isso é preciso a interação do **Back-end**, **Front-end** e dos **Bancos de dados**.

## Como uma empresa de entregas organiza rotas em tempo real?

- **Decompor o problema**: Vários entregadores, vários pedidos e destinos diferentes.
- **Reconhecer os padrões**: Horários de pico, rotas melhores, áreas com mais pedidos.
- **Abstrair o que é relevante**: Tempo estimado, distância, prioridade.
- **Criar sequências**: Para decidir quem entrega o quê, em qual ordem.

Assim, conseguimos transformar ideias em soluções. **Pensar computacionalmente** é o primeiro passo antes de qualquer linha de código.

## Decompondo problemas

Demandas surgem de forma bruta, misturando múltiplas questões e perspectivas. Para isso, precisamos decompor um problema, que é **dividi-lo em partes menores**.

Nem tudo precisa ser resolvido de uma vez. Decompor é identificar **partes** a serem resolvidas **separadamente**. **Soluções existentes** podem ser reaproveitadas para partes já conhecidas.

**Caso prático**: Melhorar rotina de estudos.
- **Subproblemas**:
  - Listar objetivos de aprendizagem
  - Identificar horários disponíveis
  - Definir prioridades e metas
  - Criar cronogramas.

**Caso prático de software**: Criar funcionalidade de login em um website.
- **Subproblemas**:
  - Identificar quais dados o sistema precisa coletar
  - Planejar como o sistema vai verificar esses dados
  - Decidir o que acontece em caso de erro
  - Planejar o que acontece após o login com êxito




## Reconhecendo padrões

Depois de decompor um problema, buscamos semelhanças entre suas diferentes partes.

**Caso prático**: Setor de atendimento ao cliente (CX)
- **Subproblemas**:
  - Muitas questões acabam se repetindo
  - **Reconhecer padrões do dia a dia** permite reutilizar respostas, automatizar e ganhar produtividade. Por exemplo, construindo um chatbot para resolver uma determinada questão para passar um passo a passo para um cliente que não entendeu alguma coisa.

### No mundo do software
Muitas funcionalidades seguem lógicas parecidas. Reutilizando ou adaptando soluções já existentes, criando estruturas reutilizáveis.



## Utilizando abstração

Por exemplo temos mapas dos principais meios de transportes em regiões metropolitanas. Nestes mapas nós geralmente temos várias linhas categorizadas por cores em uma única imagem. Nesta imagem não estão inseridas ruas, morros, ou qualquer outro tipo de informação da cidade, apenas o **passo a passo de como seguir corretamente a linha desejada.** A **ocutação** desses detalhes que "não importam" para o usuário naquele momento é um tipo de abstração.

Podemos também usar o exemplo do controle remoto. Só o que temos a disposição são os **botões que estão dispostos na interface do controle**, para que possamos apertar um botão de aumentar o volume e o volume seja aumentado. Porém, não é necessário que o usuário saiba como aquele botão está fazendo a tv aumentar o volume. **Por trás dos panos** o software do controle e da tv fazem este trabalho, que também é um tipo de abstração.

- **Abstrair é ocultar a complexidade e expor apenas o necessário**. Criando representações simplificadas e facilitando a organização e criação de grandes sistemas.

**Criando uma aplicação web**: O botão de "Enviar" é usado em várias partes com diferentes objetivos. O botão é uma abstração, a aparência e o comportamento básico estão prontos.

## Elaborando algoritmos

Algoritmo é uma sequência finita de instruções para resolver um problema. Tem começo, meio e fim. Cada passo deve ser claro, objetivo e executável. Representável em **diferentes formas**. As instruções precisam ser muito precisas para não deixar instruções ambíguas.

## Resolvendo problemas com algoritmos

**Exemplo**: Fazer um café coado.
- Objetivo: Criar um tutorial para ensinar como fazer café.
- 1º Passo: Divisão de tarefas em partes menores.
- 2º Passo: Criação do passo a passo conciso e claro.

1. Pegue um filtro de papel, um suporte e uma garrafa térmica;
2. Posicione o filtro no suporte e o suporte sobre a garrafa;
3. Adicione pó de café(1 colher por xícara de água);
4. Aqueça a água até quase ferver;
5. Derrame um pouco de água sobre o pó para umedecer;
6. Derrame o restante da água em movimentos circulares;
7. Aguarde a conclusão da filtragem;
8. Sirva o café na garrafa.

### **Caso clássico**: missionários e canibais.
- Três missionários e três canibais estão de um lado do rio e precisam atravessar.
- O barco comporta até 2 pessoas.
- Regra: nº de canibais não pode ser maior que nº de missionários nas margens. O barco não pode voltar sozinho.
- Objetivo: levar todos em segurança.

Abstração da situação:

## Representando algoritmos

Estruturando o exemplo do café:

**Início**:

Preparar utensílios:
- Pegar(filtro de papel, suporte, garrafa)
- Colocar(filtro de papel, suporte)
- Posicionar(suporte, garrafa)

Adicionar pó de café(quantidade = número de xícaras * colher):
- Adicionar(pó de café, quantidade, filtro de papel)
- Aquecer água
- Colocar(água, bule)
- Aquecer(bule, 90ºC)

Umedecer o pó de café:
- Derramar(água, modo, vazão(pouca), filtro de papel)

Filtragem principal:
- Derramar(águar, modo(circular), vazão(muita), filtro de papel)

Finalização:
- Aguardar(água, filtro de papel)
- Servir(café, garrafa)

**Fim.**

## informações importantes na hora de escrever o código:
- É melhor que criemos para adicionar aos nossos if's **funções**. Dentro dessas funções estarão as execuções a serem feitas e chamaremos elas caso uma opção if seja adequada. Para criar funções faremos **`def nome_funcao():`** com o código abaixo dela. Funções são utilizadas para deixar o código mais limpo, e quando necessário alguma alteração, podemos fazer isso apenas na função e não nas condiconais.
- Um import importante para limpar a tela é o **`import os`** que nos da a opção de utilizar o **`os.system('cls')`**.
- No pyhton podemos dizer que o arquivo é o principal, isso quer dizer que, assim que esta opção for executada ele fará uma sequência de instruções. Para informar ao python que o programa que estamos escrevendo é o principal utilizamos o seguinte formato:

```
def instrucao1():
    print('Esta é a instrução 1.')
    
def instrucao2():
    print('Esta é a instrução 2.')

def instrucao3():
    print('Esta é a instrução 3.')

def main():
    instrucao1()
    instrucao2()
    instrucao3()

if __name__ == '__main__':
    main()
```





Com o surgimento de novas versões do Python, a linguagem de programação continua a evoluir e se aprimorar, trazendo consigo uma variedade de novas funcionalidades que enriquecem a experiência de desenvolvimento. Essas inovações não apenas aprimoram a eficiência e a legibilidade do código, mas também oferecem aos desenvolvedores ferramentas mais poderosas e expressivas para lidar com desafios diversos.

Dentre as funcionalidades mais recentes, destaca-se a **introdução da instrução match no Python 3.10**, oferecendo uma abordagem mais elegante para a correspondência de padrões em dados. Essa adição não apenas **simplifica a lógica** do código, mas também proporciona uma alternativa expressiva e legível às construções tradicionais de controle de fluxo, como if elif else, que são necessários para adaptar o comportamento do programa, como vimos anteriormente.

A principal **finalidade** da instrução `match` é **simplificar a lógica** de código ao facilitar o trabalho com diferentes padrões de dados. Isso pode tornar o código mais legível e conciso em situações onde uma cadeia de if elif else pode se tornar complicada e difícil de manter.

- **Por exemplo**:
```
opcao_escolhida = int(input('Escolha uma opção: '))
if opcao_escolhida == 1:
    print('Adicionar restaurante')
elif opcao_escolhida == 2:
    print('Listar restaurantes')
elif opcao_escolhida == 3:
    print('Ativar restaurante')
elif opcao_escolhida == 4:
    print('Finalizar app')
else:
    print('Opção inválida!')
```
```
opcao_escolhida = int(input('Escolha uma opção: '))
match opcao_escolhida:
    case 1:
        print('Adicionar restaurante')
    case 2:
        print('Listar restaurantes')
    case 3:
        print('Ativar restaurante')
    case 4:
        print('Finalizar app')
    case _:
        print('Opção inválida!')
```

Perceba que recebemos a variável `opção_escolhida` como parâmetro da instrução `match` e será feito um **comparativo com todos os valores** determinados pelos blocos de case, e no final temos uma instrução case, que é um padrão curinga, que corresponde a qualquer valor que não tenha sido correspondido pelos casos anteriores, ou seja, equivalente ao else da condicional anterior.

Tendo esse exemplo, podemos entender a sintaxe padrão do match. Dentro de um bloco match, você pode utilizar a instrução case para definir padrões específicos que serão comparados com a expressão que está sendo correspondida.
```
match expressão:
    case padrão_1:
        # Código a ser executado se expressão corresponder a padrão_1
    case padrão_2:
        # Código a ser executado se expressão corresponder a padrão_2
    # ... outros casos ...
    case _:
        # Código a ser executado se nenhum dos padrões anteriores corresponder. Isso é útil para tratar casos não específicos.
```
